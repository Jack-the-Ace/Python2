## 데이터 분석 마무리 과제 - 타이타닉 데이터 분석

연습문제 데이터 파일 : titanic_kor.csv  

> 타이타닉호의 침몰은 역사상 가장 참혹한 선박 침몰 사고 중 하나이다.  
> 1912년 4월 15일, 첫 항해 중 절대 "침몰할 수 없는" 것으로 여겨졌던 RMS 타이타닉호가 빙산과 충돌한 후 침몰하게 되었다.  
> 불행히도 탑승객 모두를 태울 구명보트가 충분하지 않아 승객과 승무원 2,224명 중 1,502명이 사망하게 되었다.  
> 살아남는 데는 어느 정도 운이 작용하겠지만, 어떤 집단은 다른 집단보다 생존 가능성이 더 큰 듯 보였다.  

**아래의 승객 데이터(이름, 나이, 성별, 사회경제적 계층 등)를 사용하여  
"어떤 사람들이 생존 가능성이 더 높을까?"라는 질문에 답할 수 있는 데이터 분석을 수행해 보자.**
  
생존 : 생존여부 (0=사망, 1=생존)  
등급 : 티켓 클래스(1=1등석, 2=2등석, 3=3등석)  
성별 : 성별  
나이 : 나이  
형제자매 : 함께 탑승한 형제/자매  
부모자녀 : 함께 탑승한 부모/자녀  
요금 : 탑승 요금  
탑승지코드 : 탑승지 코드  
객실등급 : 객실 등급(1=1등석, 2=2등석, 3=3등석)  
성인여부 : 남여 성인 여부  
성인남성 : 성인 남성 여부  
선실코드 : 선실 코드  
탐승지명 : 선착장  
생존여부 : 생존 여부(no=사망, yes=생존)  
개인탑승자 : 개인 탑승자 여부

```python
#문제10. 객실 요금이 300 불 이상인 데이터에 대한 '성별', '나이', '객실등급', '생존여부' 컬럼을 출력하시오.
df.loc[df['요금'] >= 300, ['성별', '나이', '객실등급', '생존여부']]
```

```python
#문제11. '객실등급'의 데이터 값을 모두 소문자로 바꾸고 출력하시오. (First -> first, Second -> second, Third -> third)
df['객실등급'] = df['객실등급'].str.lower()
df['객실등급']
```

```python
#문제12. '성인여부' 컬럼의 데이터 값 중 man, woman 을 adult 로 바꾸시오.
df['성인여부'] = df['성인여부'].replace({'man': 'adult', 'woman': 'adult'})
df['성인여부']
```

```python
#문제13. 데이터 셋의 각 컬럼에 대한 공백(결측치)의 갯수를 출력하시오.
df.isnull().sum()
```

```python
#문제14. 데이터셋에는 총 몇 건의 데이터가 저장되어 있는지 출력하시오(전체 탑승자는 몇 명인가).
print("전체 탑승자 수:", len(df))
print("전체 탑승자 수:", df.shape) # tuple (891, 15)
print("전체 탑승자 수:", df.shape[0])
```

```python
#문제15. '생존여부' 컬럼의 데이터와 비교하여 '생존' 컬럼의 데이터 값이 동일한지를 확인하시오. 
#('생존' 컬럼의 데이터가 1 일때, '생존여부' 컬럼의 데이터가 no 이라면 서로 다른 데이터이다. 해당 데이터가 있는지를 찾으면 된다.)
mismatch = df[(df['생존'] == 1) & (df['생존여부'] == 'no')]
mismatch
# 행이 하나도 없음. 즉, 생존,생존여부 데이터의 불일치 문제는 없음.
```

```python
#문제16. '생존' 컬럼과 동일한 값이 저장되어 있는 '생존여부' 컬럼을 삭제하시오. 
#그리고 '등급' 컬럼과 동일한 값이 저장되어 있는 '객실등급' 컬럼을 삭제하시오.
df = df.drop(columns=['생존여부', '객실등급'])
df.head()
```

```python
#문제17. '탑승지코드' 컬럼과 '탑승지명' 컬럼 모두에서 데이터 값이 누락되어 있는 행(인덱스)을 찾으시오.
df[df['탑승지코드'].isnull() & df['탑승지명'].isnull()].index
```

```python
#문제18. '탑승지코드' 컬럼과 '탑승지명' 컬럼에서 결측치(데이터가 누락되어 있는)가 존재하는 행을 삭제하시오. (인덱스 61, 829 번 행에 결측치가 존재한다.)
df = df.dropna(subset=['탑승지코드', '탑승지명'])
df.shape  
# 원래 행개수 : 891 ==> 889 로 2개 제거된 것 확인.
```

```python
#문제19. 인덱스 59 ~ 63번 행에 대한 데이터만 출력하시오.
df.loc[59:63]
```

```python
#문제20. '성인여부' 컬럼에서 child 데이터에 대응하는 '나이' 컬럼의 데이터가 누락되어 있는지 확인하시오. 
#('성인여부' 컬럼의 데이터가 child 일때, '나이' 컬럼의 데이터가 isnull 인 데이터를 찾으면 된다.)
df[(df['성인여부'] == 'child') & (df['나이'].isnull())]
# 아이들 중 나이에 결측지는 존재하지 않음.
```

```python
#문제21. '나이' 컬럼의 데이터 값에 대한 평균을 출력하시오.
print("나이 평균:", df['나이'].mean())
```

```python
#문제22. '나이' 컬럼의 결측치(빈 셀)를 '나이'에 대한 평균 값으로 채우시오. 
df['나이'] = df['나이'].fillna(df['나이'].mean()) # 나이 결측값이 177개 였음.
print(df['나이'].isnull().sum())
```

```python
#문제23. '선실코드' 컬럼의 데이터 값에는 어떤 종류의 값이 있는지 출력하시오(7 개).
df['선실코드'].unique()
```

```python
#문제24. '요금' 컬럼에서 가장 높은 요금이 얼마인지 출력하시오.
print("최고 요금:", df['요금'].max())
```

```python
#문제25. 생존자에 대한 데이터만 출력하시오. ('생존' 컬럼의 데이터 값이 1 인 데이터를 추출하여 'survivor' 변수에 저장한 다음 출력하시오.)
survivor = df[df['생존'] == 1] # survivor:생존자
survivor 
# 데이터프레임 출력의 마지막에  '340 rows x 13 columns' 가 출력된 것을 확인할 수 있음. [340명의 생존자 확인]
```

```python
#문제29. '생존'과 '나이' 컬럼을 기준으로 내림차순 정렬한 다음, 20개의 데이터만 표시하시오.
df.sort_values(by=['생존', '나이'], ascending=False).head(20)
```

```python
#문제30. 여성 생존율과 남성 생존율을 아래와 같이 표시하시오.
female_survival_rate = df[df['성별'] == 'female']['생존'].mean()
male_survival_rate = df[df['성별'] == 'male']['생존'].mean()

print("여성 생존율 =", round(female_survival_rate, 2))
print("남성 생존율 =", round(male_survival_rate, 2))
```

```python
# [슬라이드로 소개 참고]

# 목표 : 승객 데이터(이름, 나이, 성별, 사회경제적 계층 등)를 사용하여 "어떤 사람들이 생존 가능성이 더 높을까?"라는 질문에 답할 수 있는 데이터 분석을 수행


# 1단계. 문제 정의
#  - 목표: 타이타닉 탑승객 중 어떤 특성을 가진 사람들이 생존할 확률이 높았는지 파악
#  - 주요 변수 후보: 성별, 나이, 객실등급, 탑승지, 요금, 동승자 수 등

# 2단계. 데이터 이해 및 전처리
#  - 결측치 처리: 나이, 탑승지코드, 선실코드 등에 결측치가 있는지 확인하고 평균/중앙값 대체 또는 삭제
#  - 데이터 변환: 범주형 → 숫자형 변환 (예: 성별 male=0, female=1).  필요 없는 중복된 컬럼 제거 (이전 과제에서 한 것처럼)
#  - 파생 변수 생성:
#    가족수 = 형제자매수 + 부모자녀수 + 1
#    연령대 = (어린이/청년/중년/노인 등)

# 3단계. 탐색적 데이터 분석 (EDA)
#  - 기초 통계: 생존/사망자 수 비율 확인
#  - 시각화:
#    성별 vs 생존율 (막대그래프)
#    객실등급 vs 생존율 (막대그래프)
#    나이 분포와 생존 여부 (히스토그램, KDE Plot)
#    요금 vs 생존 여부 (박스플롯)
#  - 교차분석:
#    성별 + 객실등급 → 생존율
#    가족 유무 → 생존율

# 4단계. 통계적 분석
#  - 성별/객실등급과 생존 여부의 관계 유의성 확인
#  - 생존자 vs 사망자의 나이 평균 비교

# 5단계. 해석 및 결론
#  - “여성, 1등석 승객, 어린이, 요금이 높은 승객, 가족이 함께 있는 경우”  → 생존 확률 70%
#  - “남성, 3등석 승객, 나이가 많은 승객” → 생존 확률 20%


# [ 확장. AI 예측 모델링 ]
#  - 로지스틱 회귀: 생존 여부에 영향을 주는 요인을 계수로 확인
#  - 의사결정나무/랜덤포레스트: 변수 중요도 확인
#  - 교차검증: 모델 성능 평가
```
