## 데이터 프레임에서 자료형을 인식하는 것은 매우 중요함.
- 연산가능 여부
- 날짜기준 분석
- 버그 예방.  "100"+"20"

```python
# 쇼핑몰 주문 기록(log) 데이터셋을 확인
import pandas as pd
df= pd.read_csv('./data_set/order_log.csv')
df
```

```python
# 주문금액의 평균을 알고싶어..
df['주문금액'].mean()
```

```python
# 주문금액은 숫자처럼 보이지만.. 숫자데이터 아님
df.info()
```

```python
# 자료형 변환 astype() -- 숫자 사이에 ,가 있어서 그냥은 못바꿈.
df['주문금액']= df['주문금액'].str.replace(',','').astype(int)  # or 'int64'
df['주문금액'].mean()
```

```python
# 주문일자 중 2월 주문내역만 추출...
# 문자열에서 2월 데이터 추출이 불편함.
df['주문일자']= pd.to_datetime(df['주문일자'])
df.dtypes
```

```python
# 2월 데이터만 추출
df[ df['주문일자'].dt.month == 2 ]
```

```python
# 연월까지 모두 필터링 2023년 4월
df[ (df['주문일자'].dt.year==2023) & (df['주문일자'].dt.month==4) ]
```

```python
# 위 and 연산을 보다 쉽게.. 하기 위해 날짜를 인덱스로 사용
df.set_index('주문일자').loc['2023-06'] #인덱스명이 일치하지 않아도 됨.
```

```python
# 결제수단 데이터 유형
df['결제수단'].value_counts()  # 각 값들의 개수
```

```python
# 결제수단 처럼 정해져있는 값을 가진 데이터의 유형을 범주형이라고 부름
df['결제수단'].dtype
```

```python
# 문자열로 되어있음.. 메모리를 조금 효율적으로 하고싶다면..
# 범주형의 자료형을 object --> category

df.info()
```

```python
df['결제수단']= df['결제수단'].astype('category')
df.info()
```
