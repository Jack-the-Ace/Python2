## 결측치
- 결측치(Missing Values)란 데이터셋에서 일부 값이 누락된 것
- 결측치가 있으면 데이터분석 결과를 왜곡할 수 있기에 분석전에 미리 전처리 해 놓아야 함.

```python
import pandas as pd
df= pd.read_csv('./data_set/pandas.csv', encoding='cp949') #제공한 csv 파일이 'cp949'로 되어 있음.
df
```

```python
# 결측치 존재 확인
df.isna() # 각 cell마다 결측치 여부를 알려줌. 데이터프레임으로 결과 줌
```

```python
df.isnull()
```

```python
# 결측치의 개수 확인
df.isna().sum()
```

```python
# 결측치의 비율이 별로 없다면 보통은 그 행을 제거한 후 분석을 함.
# 결측치의 비율 확인 - 산수
(df.isnull().mean()*100).round(2)
```

```python
# 몸무게 컴룸에서 결측치가 존재하는 행을 필터링하여 추출
df[ df['몸무게'].isna() ] #불리언 인덱싱
```

```python
# '몸무게', '급여' 컬룸의 결측치가 존재하는 행을 추출
df[  df['몸무게'].isnull() | df['급여'].isna() ]
```

```python
# 반대로 결측치가 없는 행들만 추출
df[ df['몸무게'].notna() & df['급여'].notnull() ]
```

## 결측치 처리 방법 9가지

```python
#1. 결측치가 존재하는 컬룸을 계산에서 제외하기 위해 제거 drop 
df.drop(columns= ['몸무게','급여']) #원본은 안바뀜
```

```python
# 위 방법은 몸무게와 급여 컬룸이 필요없을때..중요한 변수가 아닐때..
# 컬룸 제거 말고.. 결측치를 가진 관측값(행row)을 제거
df.drop(index=[1,3])
```

```python
# 행의 index 번호를 찾기 어려우니..결측치가 있는 행을 알아서 찾아서 제거..
df.dropna()
```

```python
#2. 0으로 채우기
df.fillna(0) #결측값을 찾아서 0으로 채워줌.
```

```python
#3. 평균값으로 채우기 -- 몸무게 컬룸만
df[['몸무게']].fillna( df['몸무게'].mean() )
```

```python
#4. 중위값 채우기 -- 급여
df[['급여']].fillna(df['급여'].median())
```

## 사례별 결측치 처리 방법 

```python
#1. 제거가 합리적인 상황 -- 설문조사 만족도 평균
df= pd.read_csv('./data_set/결측치 처리/1.survey_score.csv')
df
```

```python
#결측치 확인
df.isnull().sum()
```

```python
# 전체 데이터 대비 결측 비율 확인
(df.isnull().mean()*100).round(2)
```

```python
# 응답자 중 1명을 제거해도 전체 만족도에 큰 영향 없음.
df_drop= df.dropna()
df_drop
```

```python
# 전체 만족도 평균
print('만족도 점수:', df_drop['만족도점수'].mean().round(1))
```
