```python
# 11) '객실등급'의 데이터 값을 모두 소문자로 바꾸고 출력하시오. (First -> first, Second -> second, Third -> third)
df['객실등급']= df['객실등급'].str.lower()
df['객실등급']
```

```python
# 12) '성인여부' 컬럼의 데이터 값 중 man, woman 을 adult 로 바꾸시오.
df['성인여부']= df['성인여부'].replace({'man':'adult', 'woman':'adult'})
df['성인여부']
```

```python
# 14) 데이터셋에는 총 몇 건의 데이터가 저장되어 있는지 출력하시오(전체 탑승자는 몇 명인가)
print('전체 탑승자 수:',len(df))
print('전체 탑승자 수:',df.shape)
print('전체 탑승자 수:',df.shape[0])
```

```python
# 15) '생존여부' 컬럼의 데이터와 비교하여 '생존' 컬럼의 데이터 값이 동일한지를 확인하시오. ('생존' 컬럼의 데이터가 1
# 일때, '생존여부' 컬럼의 데이터가 no 이라면 서로 다른 데이터이다. 해당 데이터가 있는지를 찾으면 된다.)
mismatch= df[(df['생존']==1) & (df['생존여부']=='no')]
mismatch
```
