## 그래프의 종류
> 어떤 데이터셋일때 어떤 그래프가 효과적으로 변수들의 관계를 탐색할 수 있는지 확인

```python
#0. 모듈 사용
import matplotlib.pyplot as plt
```

```python
#1. 꺾은선 그래프(line plot) -- 시간에 따른 추세, 변화 확인
#예) 일별 방문자 수 변화, 월별 매출 추이, 서버트래픽 시간대별 변화...

days= [1,2,3,4,5,6,7]
visitors= [120,135,150,160,155,200,220]

plt.plot(days, visitors)
plt.title('daily visitors')
plt.xlabel('Day')
plt.ylabel('Visitors')
plt.show()
```

```python
#2. 막대 그래프 (Bar Chart) - 값 간의 크기 비교
# 요일별 평균 방문자수 비교, 상품별 판매량 비교, 부서별 성과 비교

days = [1, 2, 3, 4, 5, 6, 7]
avg_visitors = [140, 150, 170, 160, 175, 240, 220]

plt.bar(days, avg_visitors)
plt.show()
```

```python
#3. 산점도(Scatter Plot) - 두 변수 간의 관계, 분포, 이상치 확인
# 예) 광고비 vs 매출,   공부시간 vs 시험성적,   가격 vs 구매수량 (2변량 or 다변량)
ad_cost = [10, 20, 30, 40, 50, 60, 70, 80]
sales   = [100, 120, 240, 340, 360, 370, 375, 378]

plt.scatter(ad_cost, sales)
plt.show()
# 두 변수 간의 관련성이 있나?? 를 확인할때 적합. 상관관계를 볼때 용이.
```

```python
#4. 히스토그램 (histogram) - 데이터 분포, 몰림현상
# 예) 사용자 나이 분포, 주문금액 분포, 시험점수 분포 --- 일변량 그래프 (변수가 1개인 경우)
# 대부분은 중간 가격대에 몰려 있고,소수의 고가 주문이 존재하는 분포 - 전자상거래 주문 금액에서 아주 흔함
order_amounts = [
    18000, 19000, 19500, 20000, 20500, 21000, 21500, 22000,
    22500, 23000, 23500, 24000, 24500, 25000, 25500, 26000,
    26500, 27000, 27500, 28000,
    35000, 38000, 42000, 50000
]
# 가격분포의 특징 확인
plt.hist(order_amounts)  # x,y가 아님... 값만 주면 됨.
plt.show()
# 분포 특징: 2만원 대에 주문금액이 밀집. 이 회사에 대해 소비자가 생각하는 가격선... 예측
# 오른쪽으로 긴 꼬리.. -- 고가 주문은 적지만 조금씩 존재함..

# "어디에 많이 몰려있는지... 분포를 확인할때 적합"
```

```python
#5. 박스형 그래프 - 데이터들 간의 분포를 비교할때 사용.
# 예) 부서별 연봉 분포 비교, 상품 카테고리별 주문금액 비교, A/B 테스트 결과 비교, 반별 시험성적 분포 비교.

# 요일별 주문 금액 분포 비교
mon = [18000, 19000, 20000, 21000, 22000, 23000]
tue = [17000, 18000, 18500, 19000, 20000, 21000]
wed = [20000, 21000, 22000, 23000, 24000, 50000]  # 이상치 포함

plt.boxplot([mon,tue,wed])
plt.xticks([1,2,3],['mon','tue','wed'])
plt.show()

# 박스플롯이 보여주는 것.
#1. 중앙값
#2. 사분위수(Q1,Q3)
#3. 최소값, 최대값
#4. 이상치(outlier)

# 퍼져있는 분포정도를 확인

# 1) 중앙값 비교 -- 각 집단간의 크기 비교
# 2) 박스크기 비교
# 3) 수염길이 비교
# 4) 이상치 유무
```

```python
#6. 3변수 산점도
# 예) 매장별 광고비/매출액/방문자 수

# x: 광고비
# y: 매출액
#size: 방문자수 (많을수록 점의 크기가 커짐)
ad_cost = [10, 20, 30, 40, 50, 60, 70, 80]
sales   = [100, 120, 240, 340, 360, 370, 375, 378]
visitors = [150, 200, 450, 600, 850, 650, 700, 680]

plt.scatter(ad_cost, sales, s=visitors)  #s: size
plt.show()
```

```python
#7. 4변수 산점도 (x, y, size, color)
# 광고 캠페인별 성과 비교
ad_cost  = [10, 20, 30, 40, 50, 60, 70, 80]
sales    = [100, 120, 240, 340, 360, 370, 375, 378]
visitors = [150, 200, 450, 600, 700, 750, 760, 770]
campaign = [1, 1, 2, 2, 3, 3, 3, 3]  # 캠페인 유형 [1:검색광고 , 2:SNS 광고 , 3:디스플레이 광고]

plt.scatter(ad_cost, sales, s=visitors, c=campaign) # c: color
plt.show()
```
