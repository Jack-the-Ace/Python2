## 판다스의 데이터프레임을 그래프로 그리기

```python
#라이브러리 준비
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 한 카페의 월별 매출 데이터
df= pd.read_csv('./data_set/monthly_sales.csv')
df
```

```python
# 데이터프레임을 그래프로 그려보기
plt.plot(df['month'],df['sales'], marker='o', mfc='red')
#데이터가 많아질수록 좌표값이 잘 안읽어짐. 그래서 그래프에 격자표시
plt.grid(True)
plt.show()
# 월별 매출에 대한 추세... 추이.. 확인할 때 용이...
```

```python
# 어느 달에 더 많이 팔리는지 확인.. 그에 따라.. 재고관리..
# 크기비교에 적합...
plt.bar(df['month'],df['sales'])
plt.show()
```

```python
#2. pandas의 기능을 이용하여 그래프 그리기.. (feat. matplotlib)
df.plot(
    x='month',  #컬룸명
    y='sales',
    kind='line', #그래프의 종류
    title='monthly sales',
    grid=True,
    marker='o',
    markerfacecolor='yellow',
    markeredgecolor='black',
    linestyle= '--',
    linewidth=2
)
plt.show()
```

```python
#3. 다변량 그래프 (2개 이상의 컬룸.. y값이 여러개)
# 월별(x), 매출액(y1), 고객수(y2)
df= pd.read_csv('./data_set/monthly_sales_customers.csv')
df
```

```python
# 여러개의 선으로 표시..
plt.plot(df['month'],df['sales'], label='sales', marker='o')  #매출
plt.plot(df['month'],df['customers'], label='customers', marker='o')  #고객수
plt.legend()
plt.show()
```

```python
# 문제점.
# y1, y2의 단위가 다름 : 매출(만원) vs 고객수(명)
# 같은 눈금단위로 쓰는 것이 말이 안됨!

# 한 눈에 두 선을 동시에 보면서 눈금을 다르게 보려면??

fig, ax1= plt.subplots()  # 비워놓으면 1행 1열

ax1.plot(df['month'],df['sales'], marker='o', label='sales')
ax1.set_xlabel('month')
ax1.set_ylabel('sales($)')

ax2= ax1.twinx()  # x축을 공유하는 새로운 겹쳐진 그래프
ax2.plot(df['month'],df['customers'], marker='s', label='customers', color='orange')
ax2.set_ylabel('customers(person)', color='orange')
ax2.tick_params(axis='y', labelcolor='orange')

plt.show()
```
