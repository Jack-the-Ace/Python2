## Matplotlib
다양한 형태의 그래프를 통해 데이터 시각화를 할 수 있는 라이브러리

```python
import matplotlib.pyplot as plt
```

1. 그래프 기본

```python
# 꺾은선그래프 - 2차원 평면 그래프
x= [1,2,3]
y= [2,4,8]
plt.plot(x, y)
# 첫줄에 그래프의 객체값이 출력됨. 싫다면?
plt.show()
```

```python
# 그래프의 제목 설정
plt.plot(x,y)
plt.title('plot graph')
plt.show()
```

```python
# 그래프의 제목은 한글로 지정
plt.plot(x,y)
plt.title('꺾은선 그래프')
plt.show()
```

```python
# matplotlib 의 기본 글꼴이 영어글꼴이어서 한글을 인식하지 못함.
# 한글이 나오게 하려면 한글폰트를 지정해주면 됨.
import matplotlib
matplotlib.rcParams['font.family']='Malgun Gothic'  #windows
# matplotlib.rcParams['font.family']='AppleGothic'  #mac os
matplotlib.rcParams['font.size']= 15
```

```python
# 폰트의 종류를 알고싶다면.
import matplotlib.font_manager as fm
fm.fontManager.ttflist  #폰트 엔트리 객체형태로 출력되어 보기 어려움
# name 속성값만 추출
[font.name for font in fm.fontManager.ttflist]
```

```python
#한글사용
plt.plot(x,y)
plt.title('꺾은선 그래프')
plt.show()
```

```python
# 마이너스 기호를 인식하도록.
matplotlib.rcParams['axes.unicode_minus']= False
```

```python
# 한글글꼴을 사용할떄. 주의점. -기호에 대한 문제 (해결 - rcParams 지정)
x= [-1,0,1]
y= [-5,0,5]  # y=5x
plt.plot(x,y)
plt.title('y=5x 선형방정식의 그래프')
plt.show()
```

2. 축   
그래프의 축에 대한 학습

```python
x= [1,2,3]
y= [2,4,8]

plt.plot(x,y)
#축에 라벨 표시
plt.xlabel('x축', color='red', loc='right')    # left, right, center
plt.ylabel('y축', color='#00ff00', loc='top')  # top, bottom, center

#축의 눈금 지정
plt.xticks([1,2,3])
plt.yticks([3,6,9,12])
plt.show()
```

3. 범례          
그래프의 데이터선에 대한 라벨링

```python
plt.plot(x,y,label='2의 제곱')
#plt.legend(loc='lower right')  #범례를 표시해라.
plt.legend(loc=(0.6, 0.3))  #좌표 (0~1, 0~1) 범위로 x, y축 위치
plt.show()
```

5. 그래프의 크기 설정      

```python
#plt.figure( figsize=(8, 4) )  #너비, 높이
plt.figure( figsize=(8, 4), dpi=72, facecolor='yellow' ) # dot per inch,  배경색
plt.plot(x,y)
plt.show()
```

6. 그래프를 이미지 파일로 저장       

```python
plt.plot(x,y)
#plt.savefig('graph.png')  # .show() 하기 전에 저장해야만 함. show()를 한 후에 저장하면 빈 화면이 저장됨.
plt.savefig('graph2.png', dpi=48)
```

7. 텍스트 넣기

```python
plt.plot(x,y, marker='o')
plt.text(1,8,'hello')  #좌표 위치와 글씨
# 각 마커들의 위에 y값을 표시하기
for idx, txt in enumerate(y):
    plt.text(x[idx], y[idx]+0.3, txt, ha='center', color='g')
plt.show
```

8. 여러 데이터 한번에 그리기

```python
# 월별 제품별 매출 데이터 
month= ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product1=[120, 150, 170, 60, 80, 100]
product2=[90, 110, 130, 125, 140, 155]
product3=[60, 80, 75, 90, 100, 120]

# 제품별 6개월간의 판매추이를 보고싶어..
plt.plot(month, product1, label='제품1')
plt.plot(month, product2, label='제품2', marker='o', linestyle='--')
plt.plot(month, product3, label='제품3', marker='s', linestyle=':')
plt.legend(ncol=2)  # 범례 항목들이 가로로 배치...

plt.show()
```
